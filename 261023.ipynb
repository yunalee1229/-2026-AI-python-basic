```python
score = [55, 70, 85, 40, 95, 60, 75]

# basic
result1 = [s + 5 for s in score]

# if
result2 = [s for s in score if s >= 60]

# if else
result3 = ["합격" if s >= 60 else "불합격" for s in score]

print("1. 가산점 적용:", result1)
print("2. 합격 점수:", result2)
print("3. 합격 여부:", result3)
```
