```python
# -*- coding: utf-8 -*-
"""CH01 LAB 01 - 첫 프로그램: 문자열 출력하기"""

# 1. 한 줄 출력
print('반갑습니다. 파이썬입니다.')

# 2. 콤마로 여러 값 출력 (값 사이에 공백이 하나 들어간다)
print('반갑습니다.', '파이썬입니다.')

# 3. print를 두 번 쓰면 줄이 바뀐다 (위에서 아래로 실행)
print('반갑습니다.')
print('파이썬입니다.')

# 4. 작은따옴표와 큰따옴표는 똑같이 동작한다
print("큰따옴표도 됩니다.")

# 5. 따옴표 안의 따옴표
print("그는 '안녕'이라고 말했다.")
```

```python
# -*- coding: utf-8 -*-
"""CH01 LAB 02 - print 함수의 여러 사용법"""

# 문자열 반복: * 연산자
print('=' * 30)
print('파이썬 마스터 CH01')
print('=' * 30)

# 빈 줄 출력
print()

# 문자열 연결: + 연산자
print('Hi! ' + 'Python')

# 따옴표 유무의 차이
print(3 + 5)        # 8      <- 수식은 계산된다
print('3 + 5')      # 3 + 5  <- 문자열은 그대로 출력된다

# 줄바꿈 문자 \n
print('첫째 줄\n둘째 줄')

# 탭 문자 \t
print('이름\t학과')
print('홍길동\t소프트웨어학과')
```

```python
# -*- coding: utf-8 -*-
"""CH01 LAB 03 - 산술 연산 결과 출력하기"""

print('덧셈:', 15 + 4)
print('뺄셈:', 15 - 4)
print('곱셈:', 15 * 4)
print('나눗셈:', 15 / 4)       # 결과는 항상 실수(float)
print('몫:', 15 // 4)
print('나머지:', 15 % 4)
print('거듭제곱:', 15 ** 2)

print()
print('-- / 와 // 의 차이 --')
print('15 / 3  =', 15 / 3)     # 5.0  (실수)
print('15 // 3 =', 15 // 3)    # 5    (정수)

print()
print('-- 연산자 우선순위 --')
print('2 + 3 * 4   =', 2 + 3 * 4)     # 14  (* 가 먼저)
print('(2 + 3) * 4 =', (2 + 3) * 4)   # 20  (괄호가 먼저)
```

```python
# -*- coding: utf-8 -*-
"""CH01 LAB 04 - 자기소개 카드 만들기 (종합)

CH01에서 배운 내용만 사용한다: print, 문자열, 문자열 반복/연결, 산술 연산.
"""

print('+' + '-' * 34 + '+')
print('|' + ' ' * 11 + '자기소개 카드' + ' ' * 10 + '|')
print('+' + '-' * 34 + '+')
print('| 이름   : 홍길동                  |')
print('| 학과   : 소프트웨어학과          |')
print('| 학번   : 202612345               |')
print('| 취미   : 코딩, 독서              |')
print('+' + '-' * 34 + '+')

print()
# 태어난 해로 나이 계산하기 (수식은 계산되어 출력된다)
print('2026년 기준 나이:', 2026 - 2007, '세')
print('지금까지 산 개월 수(대략):', (2026 - 2007) * 12, '개월')
```
