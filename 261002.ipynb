```python
# 논리 연산자: 참과 거짓의 논리 동작을 다루는 연산자
```

```python
import random

com = random.randint(1, 3)
user = int(input('가위(1), 바위(2), 보(3) 중 선택: '))

if com == 1:
    print('컴퓨터: 가위')
elif com == 2:
    print('컴퓨터: 바위')
else:
    print('컴퓨터: 보')

if user == com:
    print('비김')
elif user == 1 and com == 3:
    print('이김 ㅊㅊ')
elif user == 2 and com == 1:
    print('이김 ㅊㅊ')
elif user == 3 and com == 2:
    print('이김 ㅊㅊ')
else:
    print('짐')
```

```python
import random

com = random.randint(1, 100)
user = int(input('1부터 100까지의 자연수 중 임의의 수 하나를 입력하세요: '))

print(com)
print(user)

if com > user:
    print('컴퓨터의 수보다 작은 수입니다.')
elif com < user:
    print('컴퓨터의 수보다 큰 수입니다.')
else:
    print('컴퓨터의 수와 같습니다.')
```

```python
x = int(input('첫 번째 수: '))
y = int(input('두 번재 수: '))
op = input('연산자를 입력하세요')
def asmd(x, y, op):
    if op == '+':
        return x + y
    elif op == '-':
        return x - y
    elif op == '*':
        return x * y
    elif op == '/':
        return x / y
    else:
        print('잘못된 연산자입니다')
asmd(x, y, op)
```

```python
# break: 조건식과 관계없이 즉시 반복문을 빠져나오는 기능. 보통 if 문과 함께 특정 구간까지 실행 후 break로 마침
```

```python
# while 문: 조건식이 참인 동안 ?
import random
com = random.randint(1, 100)
while True:
    user = int(input('당신이 생각하는 수: '))
    print(com)
    if user == com:
        print('정답')
        break
    elif user > com:
        print('입력한 숫자가 큽니다')
    else:
        print('입력한 숫자가 작습니다')
```
