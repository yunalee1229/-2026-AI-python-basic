```python
# 주문 금액 계산

americano = 2000
latte = 3000
tea = 2500

total = 0 # 합계를 0으로 초기화

a_num = int(input('아메리카노 수량: '))
total += a_num * americano

l_num = int(input('라떼 수량: '))
total += l_num * latte

t_num = int(input('티 수량: '))
total += t_num * tea

print('-'*30)
print(f'{total:,}원')
print('-'*30)
```

```python
sub1 = int(input('1과목 점수: '))
sub2 = int(input('2과목 점수: '))
sub3 = int(input('3과목 점수: '))

avg = (sub1 + sub2 + sub3) // 3

print('Loading Complete')

print(f'{avg:.2f}')

if avg >= 60:
    print('합격')
else:
    print('불합격')
```

```python
# 숫자를 입력받아 짝수인지 홀수인지 출력

num = int(input("계산할 숫자를 입력하세요: "))
res = num % 2
if res == 1:
    print('홀수')
else:
    print('짝수')
```
